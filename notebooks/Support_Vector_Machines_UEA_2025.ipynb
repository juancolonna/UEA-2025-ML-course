### Ajustando o SVM: suavizando margens

Nossa discussão até agora se concentrou em conjuntos de dados muito limpos, nos quais existe um limite de decisão perfeito.

Mas e se seus dados tiverem alguma sobreposição? Por exemplo, você pode ter dados como este:

```python
# X, y = make_blobs(n_samples=100, centers=2, random_state=0, cluster_std=1.2)
# fig = plt.figure(figsize=(8,7))
# plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn');
```

Para lidar com esse caso, a implementação do SVM usaremos o C e diferentes kernels.

- Para $C$ muito grandes, a margem é "intolerante" e os pontos não podem atravesá-la.

- Para $C$ menores, a margem é mais tolerante, e pode abranger alguns pontos.
